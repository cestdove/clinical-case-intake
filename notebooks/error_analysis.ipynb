## Error Analysis

1. First of all importing all and setting up the new error dataframe.

```python
import pandas as pd 
import joblib 
```

```python
df_test = pd.read_csv("../datasets/clinical_cases_test.csv")    # importing the test set

# importing models
model = joblib.load("../models/clinical_classifier.joblib")
vectorizer = joblib.load("../models/tfidf_vectorizer.joblib")
```

```python
X_test = df_test["medical_abstract"]

# vectorizing the abstracts
X_test_vec = vectorizer.transform(X_test)
```

```python
predictions = model.predict(X_test_vec)     # obtaining the predictions
```

```python
# starting error analysis creating a new copy of original df but where predictions are all wrong
errors = df_test[predictions != df_test["condition_label"]].copy()

errors["predicted_label"] = predictions[predictions != df_test["condition_label"]]
```

```python
errors.shape    # errors count
```

2. Starting to work with errors to analyze the patterns.

```python
# visualizing full text rows to get better reasoning of vectorizer errors 
pd.set_option("display.max_colwidth", None)
```

```python
errors[[
    "medical_abstract",
    "condition_label",
    "predicted_label"
]].head(10)
```

The results clearly show how the model isn't doing necessarily bad, besides the low 50% accuracy trend.

Most medical abstracts contain multiple words from multiple topics, about the different labels we have. 
It would be better to add maybe more labels to the downloaded dataset we got, or search already for better technologies, like transformers. 

Surely the lack of context reading from the model is a difficult flaw, but let's deep dive through the complete set first to see the full trend on the errors.

```python
# random 10 samples from the set
errors[[
    "medical_abstract",
    "condition_label",
    "predicted_label"
]].sample(10, random_state=42)  # remove random_state=42 to check different results for each execution
```

```python
model.coef_.shape   # number of rows and columns (class, feature) mantained as coef by the vectorizer
```

```python
feature_names = vectorizer.get_feature_names_out()  # extracting the feature names to link them soon

feature_names.shape
```

```python
class_1_coefficients = model.coef_[0]   # first class coeffs 

top_indices = class_1_coefficients.argsort()[-10:][::-1]   # orders by value, takes top 10 in desc order 

for index in top_indices:
    print(feature_names[index], class_1_coefficients[index]) # prints in order token and weight value 
```

Doing the same for the 5th class where the model makes more mistakes.

```python
class_5_coefficients = model.coef_[4] # 5th class

top_indices = class_5_coefficients.argsort()[-10:][::-1]   # orders by value, takes top 10 in desc order 

for index in top_indices:
    print(feature_names[index], class_5_coefficients[index]) # prints in order token and weight value 
```
